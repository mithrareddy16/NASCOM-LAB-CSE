```python
import pandas as pd
data = {
    "Student": ["S1", "S2", "S3", "S4", "S5", "S6", "S7", "S8", "S9", "S10"],
    "Study_Hours": [2, 3, 4, 5, 6, 7, 8, 4, 6, 9],
    "Attendance": [65, 70, 75, 80, 85, 90, 92, 78, 88, 95],
    "Assignment_Score": [55, 60, 65, 72, 78, 85, 88, 70, 80, 92],
    "Previous_Score": [58, 62, 68, 70, 76, 82, 86, 72, 79, 90],
    "Final_Score": [60, 64, 70, 75, 82, 88, 91, 74, 84, 94]
}
df = pd.DataFrame(data)
print("Complete Dataset:")
print(df)
print("\nAverage Study Hours:", df["Study_Hours"].mean())
print("Average Attendance:", df["Attendance"].mean())
print("Average Assignment Score:", df["Assignment_Score"].mean())
print("Average Previous Score:", df["Previous_Score"].mean())
print("Average Final Score:", df["Final_Score"].mean())
print("\nStudent with highest Final Score:")
print(df.loc[df["Final_Score"].idxmax()])
print("\nStudent with lowest Final Score:")
print(df.loc[df["Final_Score"].idxmin()])
print("\nStudents with Attendance greater than 80%:")
print(df[df["Attendance"] > 80])
print("\nStudents with Final Score greater than 80:")
print(df[df["Final_Score"] > 80])
print("\nStatistical Summary:")
print(df.describe())
```

```python
import numpy as np
final_scores = df["Final_Score"].to_numpy()
print("Final Scores:", final_scores)
print("Mean:", np.mean(final_scores))
print("Median:", np.median(final_scores))
print("Maximum:", np.max(final_scores))
print("Minimum:", np.min(final_scores))
print("Standard Deviation:", np.std(final_scores))
```

```python
import pandas as pd
import numpy as np
data = {
    "Student": ["S1", "S2", "S3", "S4", "S5", "S6", "S7", "S8", "S9", "S10"],
    "Study_Hours": [2, 3, 4, 5, 6, 7, 8, 4, 6, 9],
    "Attendance": [65, 70, 75, 80, 85, 90, 92, 78, 88, 95],
    "Assignment_Score": [55, 60, 65, 72, 78, 85, 88, 70, 80, 92],
    "Previous_Score": [58, 62, 68, 70, 76, 82, 86, 72, 79, 90],
    "Final_Score": [60, 64, 70, 75, 82, 88, 91, 74, 84, 94]
}
df = pd.DataFrame(data)
X = df[["Study_Hours", "Attendance", "Assignment_Score"]].values
Y = df["Final_Score"].values
X = np.column_stack((np.ones(len(X)), X))
beta = np.linalg.inv(X.T @ X) @ X.T @ Y
print("Coefficients:")
print(beta)
print("\nIntercept (b0):", beta[0])
print("Study Hours coefficient (b1):", beta[1])
print("Attendance coefficient (b2):", beta[2])
print("Assignment Score coefficient (b3):", beta[3])
```
